```python
arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([2, 4, 5, 8])

print("Addition:", arr1 + arr2)
print("Subtraction:", arr1 - arr2)
print("Multiplication:", arr1 * arr2)
print("Division:", arr1 / arr2)
```

```python
arr = np.array([1, 2, 3, 4, 5])

print("Square:", arr ** 2)
print("Cube:", arr ** 3)
```

```python
arr1 = np.array([10, 50, 30, 80])
arr2 = np.array([20, 40, 35, 60])

print("Maximum:", np.maximum(arr1, arr2))
print("Minimum:", np.minimum(arr1, arr2))
```

```python
arr = np.arange(21)

print("Array:", arr)

# First five elements
print("First five:", arr[:5])

# Last five elements
print("Last five:", arr[-5:])

# Index 5 to 15
print("Index 5 to 15:", arr[5:16])

# Every second element
print("Every second element:", arr[::2])
```
