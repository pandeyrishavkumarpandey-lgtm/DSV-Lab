```python
import numpy as np
import pandas as pd
```

```python
arr = np.arange(11, 30, 2)
print(arr)
print('Dimensions:', arr.ndim)
print('Elements:', arr.size)
```

```python
values = np.linspace(5, 35, 7)
print(values)
```

```python
arr = np.full((4, 3), 9)
arr[2, 1] = 0
print(arr)
```

```python
a = np.array([6, 12, 18, 24])
print(a % 5)
```

```python
arr = np.array([14, 28, 42, 56, 70, 84])
print(arr[1::2])
```

```python
arr = np.array([-16, 25, -36, 49])
absolute = np.abs(arr)
print('Absolute:', absolute)
print('Square roots:', np.sqrt(absolute))
```

```python
matrix = np.arange(7, 13).reshape(3, 2)
print(matrix)
print('Original shape:', matrix.shape)
print('Transpose:\n', matrix.T)
print('Transposed shape:', matrix.T.shape)
```

```python
temperatures = np.array([18, 22, 27, 31, 24, 19])
print('Range:', temperatures.max() - temperatures.min())
print('Median:', np.median(temperatures))
```

```python
values = np.array([4, 9, 4, 2, 9, 7, 2, 4])
unique_values = np.unique(values)
print(unique_values)
print('Count:', unique_values.size)
```

```python
speed = np.array([35, 62, 48, 81, 55])
labels = np.where(speed >= 60, 'Fast', 'Normal')
print(labels)
```

```python
books_issued = pd.Series([24, 31, 19, 27], index=['Monday', 'Tuesday', 'Wednesday', 'Thursday'])
print(books_issued)
```

```python
prices = pd.Series([45000, 18000, 22000], index=['Laptop', 'Tablet', 'Phone'])
print(prices['Tablet'])
```

```python
rainfall = pd.Series([12, 0, 18, 5, 23])
print(rainfall[rainfall > 10])
```

```python
prices = pd.Series({'Gold': 7200, 'Silver': 95, 'Copper': 850})
print('Index:', prices.index)
print('Values:', prices.values)
```

```python
ratings = pd.Series([3.5, 4.2, 2.8, 4.7])
ratings.name = 'Customer Ratings'
ratings.index.name = 'Review ID'
print(ratings)
```

```python
s = pd.Series([11, 22, 33], index=['X1', 'X2', 'X3'])
print('X2 exists:', 'X2' in s.index)
print('X4 absent:', 'X4' not in s.index)
```

```python
stock = pd.Series([16, 28, 12], index=['Keyboard', 'Mouse', 'Monitor'])
print(stock.to_dict())
```

```python
marks = {'Hindi': 78, 'English': 84, 'Computer': 91}
s = pd.Series(marks, index=['Computer', 'Hindi', 'Art', 'English'])
print(s)
print('Missing entry:', s[s.isnull()].index.tolist())
```

```python
expenses = pd.Series([1200, None, 850, None, 600])
print('isnull():\n', expenses.isnull())
print('notnull():\n', expenses.notnull())
```

```python
a = pd.Series([8, 15, 22], index=['P', 'Q', 'R'])
b = pd.Series([4, 6, 10], index=['R', 'P', 'Q'])
print(a + b)
```
